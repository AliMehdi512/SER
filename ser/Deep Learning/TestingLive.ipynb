```python
"""
This file can be used to try a live prediction. 
"""

import keras
import numpy as np
import librosa

class livePredictions:
    """
    Main class of the application.
    """

    def __init__(self, path, file):
        """
        Init method is used to initialize the main parameters.
        """
        self.path = path
        self.file = file

    def load_model(self):
        """
        Method to load the chosen model.
        :param path: path to your h5 model.
        :return: summary of the model with the .summary() function.
        """
        self.loaded_model = keras.models.load_model(self.path)
        return self.loaded_model.summary()

    def makepredictions(self):
        """
        Method to process the files and create your features.
        """
        data, sampling_rate = librosa.load(self.file)
        mfccs = np.mean(librosa.feature.mfcc(y=data, sr=sampling_rate, n_mfcc=40).T, axis=0)
        x = np.expand_dims(mfccs, axis=1)
        x = np.expand_dims(x, axis=0)
        pred = self.loaded_model.predict(x)
        predictions = np.argmax(pred, axis=1)[0]
        print("Prediction is", " ", self.convertclasstoemotion(predictions))

    @staticmethod
    def convertclasstoemotion(pred):
        """
        Method to convert the predictions (int) into human readable strings.
        """
        
        label_conversion = {'0': 'neutral',
                            '1': 'calm',
                            '2': 'happy',
                            '3': 'sad',
                            '4': 'angry',
                            '5': 'fearful',
                            '6': 'disgust',
                            '7': 'surprised'}

        return label_conversion.get(str(pred), "Unknown")

# Here you can replace path and file with the path of your model and of the file 
# from the RAVDESS dataset you want to use for the prediction.

pred = livePredictions(path='testing-16-200_model.h5', file='../examples/1.mp3')

pred.load_model()
pred.makepredictions()
```

```python
# Live Prediction
import keras
import numpy as np
import librosa

class livePredictions:
    def __init__(self, path, file):
        self.path = path
        self.file = file

    def load_model(self):
        self.loaded_model = keras.models.load_model(self.path)
        return self.loaded_model.summary()

    def makepredictions(self, interval_sec=2):
        """
        Predict emotion for each audio interval (in seconds).
        """
        data, sampling_rate = librosa.load(self.file)
        total_duration = librosa.get_duration(y=data, sr=sampling_rate)
        samples_per_interval = int(sampling_rate * interval_sec)

        print(f"\nTotal duration: {total_duration:.2f} seconds")
        print(f"Processing in intervals of {interval_sec} seconds...\n")

        for start in range(0, len(data), samples_per_interval):
            end = start + samples_per_interval
            segment = data[start:end]

            # Pad if the segment is shorter than expected (last chunk)
            if len(segment) < samples_per_interval:
                segment = np.pad(segment, (0, samples_per_interval - len(segment)))

            mfccs = np.mean(librosa.feature.mfcc(y=segment, sr=sampling_rate, n_mfcc=40).T, axis=0)
            x = np.expand_dims(mfccs, axis=1)
            x = np.expand_dims(x, axis=0)

            pred = self.loaded_model.predict(x, verbose=0)
            prediction = np.argmax(pred, axis=1)[0]
            emotion = self.convertclasstoemotion(prediction)

            start_sec = start / sampling_rate
            end_sec = min(end / sampling_rate, total_duration)
            print(f"[{start_sec:.2f}s - {end_sec:.2f}s]: {emotion}")

    @staticmethod
    def convertclasstoemotion(pred):
        label_conversion = {
            '0': 'neutral',
            '1': 'calm',
            '2': 'happy',
            '3': 'sad',
            '4': 'angry',
            '5': 'fearful',
            '6': 'disgust',
            '7': 'surprised'
        }
        return label_conversion.get(str(pred), "Unknown")
```

```python
# Choose path to model and file
pred = livePredictions(path='testing-16-200_model.h5', file='../examples/zb.mpga')

# Load model
pred.load_model()

# Predict with 3-second intervals
pred.makepredictions(interval_sec=3)
```
